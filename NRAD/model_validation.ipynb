```python
import matplotlib.pyplot as plt
import numpy as np
import os
```

```python
def plot_loss(epochs, train_loss, val_loss, title):
    plt.figure(figsize=(10, 5))
    plt.plot(epochs, train_loss, label="Train Loss", marker='o', linestyle='-', color='blue')
    plt.plot(epochs, val_loss, label="Validation Loss", marker='s', linestyle='--', color='orange')

    plt.title(f"{title}")
    plt.xlabel("Epochs")
    plt.ylabel("Loss")

    plt.legend()
    plt.grid(alpha=0.3)
    plt.tight_layout()
    plt.show()
```

```python
seed = 2
Loss_Data = {}
for i in range(1, 11):
    loss_path = f"models/loss_data/context_weight_MC{seed:02d}_Data{i:02d}_loss.npz"
    Loss_Data[f"context_weights_MC{seed:02d}_Data{i:02d}"] = np.load(loss_path, allow_pickle=True)
    plot_loss(Loss_Data[f"context_weights_MC{seed:02d}_Data{i:02d}"]["epochs"], Loss_Data[f"context_weights_MC{seed:02d}_Data{i:02d}"]["train_loss"], Loss_Data[f"context_weights_MC{seed:02d}_Data{i:02d}"]["val_loss"], f"context_weights_MC{seed:02d}_Data{i:02d}")
```

```python
model = "reweight"
for i in range(1, 11):
    loss_path = f"models/loss_data/{model}_MC{seed:02d}_Data{i:02d}_loss.npz"
    Loss_Data[f"{model}_MC{seed:02d}_Data{i:02d}"] = np.load(loss_path, allow_pickle=True)
    plot_loss(Loss_Data[f"{model}_MC{seed:02d}_Data{i:02d}"]["epochs"], Loss_Data[f"{model}_MC{seed:02d}_Data{i:02d}"]["train_loss"], Loss_Data[f"{model}_MC{seed:02d}_Data{i:02d}"]["val_loss"], f"{model}_MC{seed:02d}_Data{i:02d}")    
```
