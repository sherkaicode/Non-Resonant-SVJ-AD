```python
import glob
import uproot
import os
from collections import defaultdict
```

```python
Diboson_files = glob.glob("/home/aegis/ether/Research_HEP/Dataset_ver3/MC/reduce_root/Multijet/*/root_*.root")
```

```python
# Dictionary to store counts per subprocess
event_counts = defaultdict(int)

# print(f"{'Subprocess':<40} | {'File Name':<15} | {'Events':<10}")
# print("-" * 75)

for file_path in Diboson_files:
    # 1. Identify Subprocess Name (Parent Directory)
    # Extracts 'mc20_13TeV_MC_Sh_2211_WlvZqq' from the path
    subprocess_name = os.path.basename(os.path.dirname(file_path))
    
    # 2. Open file and count events
    try:
        with uproot.open(file_path) as file:
            # Try to find the tree name automatically
            # Common names: 'nominal', 'events', 'CollectionTree'
            tree_name = None
            
            # Check for 'nominal' first (Standard ATLAS)
            if "nominal" in file:
                tree_name = "nominal"
            else:
                # Fallback: grab the first TTree found
                for key, obj in file.items():
                    if isinstance(obj, uproot.behaviors.TTree.TTree):
                        tree_name = key
                        break
            
            if tree_name:
                count = file[tree_name].num_entries
                event_counts[subprocess_name] += count
                # print(f"{subprocess_name:<40} | {os.path.basename(file_path):<15} | {count:<10}")
            else:
                print(f"Warning: No TTree found in {os.path.basename(file_path)}")

    except Exception as e:
        print(f"Error reading {file_path}: {e}")

print("-" * 75)
print("\nTotal Event Counts per Subprocess:")
print("=" * 50)
for sub, total in event_counts.items():
    print(f"{sub}: {total}")
```

```python
data_files = glob.glob("/home/aegis/ether/Research_HEP/Dataset_ver2/Data/reduce_root/*/run*/root_*.root")

import uproot
import glob
import os

# 1. Get the list of files using the pattern you provided
path_pattern = "/home/aegis/ether/Research_HEP/Dataset_ver2/Data/reduce_root/*/run*/root_*.root"
data_files = glob.glob(path_pattern)

print(f"Found {len(data_files)} files. Calculating total events...")

total_events = 0

# 2. Loop through files and sum the events
for file_path in data_files:
    try:
        with uproot.open(file_path) as file:
            # Attempt to find the main tree (usually 'nominal' or the first available tree)
            if "nominal" in file:
                count = file["nominal"].num_entries
            else:
                # Fallback: grab the first tree found in the file
                keys = file.keys(cycle=False) # cycle=False removes ;1 versions
                first_tree = next((k for k in keys if isinstance(file[k], uproot.behaviors.TTree.TTree)), None)
                
                if first_tree:
                    count = file[first_tree].num_entries
                else:
                    count = 0
            
            total_events += count

    except Exception as e:
        print(f"Error reading {os.path.basename(file_path)}: {e}")

print("=" * 30)
print(f"TOTAL EVENTS: {total_events}")
print("=" * 30)
```
