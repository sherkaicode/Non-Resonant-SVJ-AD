```python
df_all_MC = pd.read_csv("Full_MC_Data.csv")
```

```python
df_all_MC
```

```python
N_total = len(df_all_MC)
chunk_size = 10_000_000
n_chunks = math.ceil(N_total / chunk_size)

print(f"Splitting into {n_chunks} chunks of {chunk_size:,} each.")
```

```python
df_all_MC_shuffled = df_all_MC.sample(frac=1, random_state=42).reset_index(drop=True)

columns = list(df_all_MC_shuffled.columns)

for i in range(n_chunks):
    start = i * chunk_size
    end = min((i + 1) * chunk_size, N_total)
    print(f"Saving chunk {i+1}/{n_chunks}: rows {start:,}–{end:,}")

    df_chunk = df_all_MC_shuffled.iloc[start:end]
    df_chunk = df_chunk.astype("float32", errors="ignore")

    arrays = {col: df_chunk[col].to_numpy() for col in columns}

    np.savez_compressed(f"chunks_MC/mc_chunk_{i+1:02d}.npz", **arrays)
```
