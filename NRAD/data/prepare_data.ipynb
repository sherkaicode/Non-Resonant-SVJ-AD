```python
import pandas as pd
import os
from tqdm import tqdm  # optional for progress bar
from glob import glob
import numpy as np
import math
```

```python
df_all = pd.read_csv("Full_Data.csv")
```

```python
N_total_data = len(df_all)              # 274_816_956
chunk_size = 10_000_000
n_chunks = math.ceil(N_total_data / chunk_size)

print(f"Splitting into {n_chunks} chunks of {chunk_size:,} each.")
```

```python
df_all_shuffled = df_all.sample(frac=1, random_state=42).reset_index(drop=True)

columns = list(df_all_shuffled.columns)

for i in range(n_chunks):
    start = i * chunk_size
    end = min((i + 1) * chunk_size, N_total_data)
    print(f"Saving chunk {i+1}/{n_chunks}: rows {start:,}–{end:,}")

    df_chunk = df_all_shuffled.iloc[start:end]
    df_chunk = df_chunk.astype("float32", errors="ignore")

    arrays = {col: df_chunk[col].to_numpy() for col in columns}

    np.savez_compressed(f"chunks/data_chunk_{i+1:02d}.npz", **arrays)
```
