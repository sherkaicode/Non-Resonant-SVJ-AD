```python
def sr_mask(events):
    ht = events[:,0]
    met = events[:,1]
    # Apply SR cuts
    mask = (ht > 600) & (met > 600)
    return mask


sr_mask = (df_all_MC['ht'] > 600) & (df_all_MC['met'] > 600)

df_SR = df_all_MC[sr_mask]
```
