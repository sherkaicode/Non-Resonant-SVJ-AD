```python
import pandas as pd
pd.set_option('display.max_colwidth', None)
```

```python
Diboson = pd.read_csv("MC_Events/MC_Diboson.csv")
Multijet = pd.read_csv("MC_Events/MC_Multijet.csv")
Wjets = pd.read_csv("MC_Events/MC_Wjets.csv")
Zjets = pd.read_csv("MC_Events/MC_Zjets.csv")
ttbar = pd.read_csv("MC_Events/MC_ttbar.csv")
single_top = pd.read_csv("MC_Events/MC_Single_top.csv")
```

```python
print("Diboson Events:", Diboson.shape[0])
print("Multijet Events:", Multijet.shape[0])
print("Wjets Events:", Wjets.shape[0])
print("Zjets Events:", Zjets.shape[0])
print("ttbar Events:", ttbar.shape[0])
print("Single_top Events:", single_top.shape[0])
```

```python
Diboson_summary = Diboson.groupby('source').agg(
    count=('source', 'size'),
    total_weight=('weight', 'sum')
).reset_index()

Multijet_summary = Multijet.groupby('source').agg(
    count=('source', 'size'),
    total_weight=('weight', 'sum')
).reset_index()
Wjets_summary = Wjets.groupby('source').agg(
    count=('source', 'size'),
    total_weight=('weight', 'sum')
).reset_index()
Zjets_summary = Zjets.groupby('source').agg(
    count=('source', 'size'),
    total_weight=('weight', 'sum')
).reset_index()
ttbar_summary = ttbar.groupby('source').agg(
    count=('source', 'size'),
    total_weight=('weight', 'sum')
).reset_index()
single_top_summary = single_top.groupby('source').agg(
    count=('source', 'size'),
    total_weight=('weight', 'sum')
).reset_index()

print(Diboson_summary)
print(Multijet_summary)
print(Wjets_summary)
print(Zjets_summary)
print(ttbar_summary)
print(single_top_summary)
```

```python
print("Diboson:", sum(Diboson['weight']))
print("Multijet:", sum(Multijet['weight']))
print("W+jets:", sum(Wjets['weight']))
print("Z+jets:", sum(Zjets['weight']))
print("ttbar:", sum(ttbar['weight']))
print("Single top:", sum(single_top['weight']))
```

```python
def sr_mask(events):
    ht = events[:,0]
    met = events[:,1]
    # Apply SR cuts
    mask = (ht > 600) & (met > 600)
    return mask
```

```python
sr_mask_Diboson = (Diboson['ht'] > 600) & (Diboson['met'] > 600)
sr_mask_Multijet = (Multijet['ht'] > 600) & (Multijet['met'] > 600)
sr_mask_Wjets = (Wjets['ht'] > 600) & (Wjets['met'] > 600)
sr_mask_Zjets = (Zjets['ht'] > 600) & (Zjets['met'] > 600)
sr_mask_ttbar = (ttbar['ht'] > 600) & (ttbar['met'] > 600)
sr_mask_single_top = (single_top['ht'] > 600) & (single_top['met'] > 600)   
```

```python
print("Diboson SR:", sum(Diboson['weight'][sr_mask_Diboson]))
print("Multijet SR:", sum(Multijet['weight'][sr_mask_Multijet]))
print("W+jets SR:", sum(Wjets['weight'][sr_mask_Wjets]))
print("Z+jets SR:", sum(Zjets['weight'][sr_mask_Zjets]))
print("ttbar SR:", sum(ttbar['weight'][sr_mask_ttbar]))
print("Single top SR:", sum(single_top['weight'][sr_mask_single_top]))
```

```python
print("Count of events in SR:")
print("Diboson SR count:", sr_mask_Diboson.sum())
print("Multijet SR count:", sr_mask_Multijet.sum())
print("W+jets SR count:", sr_mask_Wjets.sum())
print("Z+jets SR count:", sr_mask_Zjets.sum())
print("ttbar SR count:", sr_mask_ttbar.sum())
```

```python
print("Count of events in CR:")
print("Diboson CR count:", (~sr_mask_Diboson).sum())
print("Multijet CR count:", (~sr_mask_Multijet).sum())
print("W+jets CR count:", (~sr_mask_Wjets).sum())
print("Z+jets CR count:", (~sr_mask_Zjets).sum())   
print("ttbar CR count:", (~sr_mask_ttbar).sum())
print("Single top CR count:", (~sr_mask_single_top).sum())
```
