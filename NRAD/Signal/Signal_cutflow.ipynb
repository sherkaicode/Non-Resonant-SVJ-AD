```python
# nc_files = os.listdir('detector_signals_NC')
nc_files = os.listdir('replicate_deepak_nc')
df_signals_nc = {}

for file in nc_files:
    if file.endswith('.txt'):
        file_name = file.replace('.txt', '')
        mass = file_name.split('_')[0]
        rinv = file_name.split('_')[1]
        if not (rinv == '8'):
            continue
        detector = "-".join(file_name.split('_')[5:])
        key = f"{mass}_{rinv}_{detector}"
        df_signals_nc[key] = pd.read_csv(os.path.join('detector_signals_NC', file), delim_whitespace=True)
        print("Number of events in", key, ":", sum(df_signals_nc[key]["weight"]))
```

```python
# files = os.listdir('detector_signals')
files = os.listdir('replicate_deepak')

# print(files)
df_signals = {}

for file in files:
    if file.endswith('.txt'):
        file_name = file.replace(".txt", "")
        mass = file_name.split('_')[0]
        rinv = file_name.split('_')[1]
        if not (rinv == '8'):
            continue
        detector = "-".join(file_name.split('_')[5:])

        df_signals[f"{mass}-{rinv}-{detector}"] = pd.read_csv(os.path.join('detector_signals', file), delim_whitespace=True)
        print("Number of events for", mass, rinv, detector, ":", sum(df_signals[f"{mass}-{rinv}-{detector}"]["weight"]))
```

```python
for key in df_signals:
    mask = (df_signals[key]['ht'] > 600) & (df_signals[key]['met'] > 600)   
    print(f"Number of events passing ht>600 and met>600 for {key}:", sum(df_signals[key][mask]['weight']))
```

```python
import pandas as pd
import os
import numpy as np

# Results list to store each row
results = []

# Assuming df_signals_nc and df_signals are already populated from your previous code blocks
for key_1st in df_signals.keys():
    # 1. Standardize the key to match the Initial dictionary (mass_rinv_detector)
    # df_signals uses mass-rinv-detector, while nc uses mass_rinv_detector
    key_nc = key_1st.replace('-', '_', 2) 
    
    # 2. Calculate sums
    # Initial (from df_signals_nc)
    initial_val = sum(df_signals_nc[key_nc]["weight"]) if key_nc in df_signals_nc else 0
    
    # 1st Cut (from df_signals)
    cut1_val = sum(df_signals[key_1st]["weight"])
    
    # 2nd Cut (Manual mask)
    mask = (df_signals[key_1st]['ht'] > 600) & (df_signals[key_1st]['met'] > 600)
    cut2_val = sum(df_signals[key_1st][mask]['weight'])
    
    # 3. Append to results
    results.append({
        'Detector': key_1st,
        'Initial': initial_val,
        '1st Cut': cut1_val,
        '2nd Cut': cut2_val,
        'Efficiency after 1st Cut (%)': (cut1_val / initial_val * 100) if initial_val > 0 else 0,
        'Efficiency after 2nd Cut (%)': (cut2_val / cut1_val * 100) if cut1_val > 0 else 0,
        'Cross Section': 9551,
        'Expected Events': 9551*139, 
        'Events after 1st Cut': (cut1_val / initial_val * 9551*139) if initial_val > 0 else 0,
        'Events after 2nd Cut': (cut2_val / initial_val * 9551*139) if initial_val > 0 else 0,
        'Significance (S/sqrt(B))': ((cut2_val / initial_val * 9551*139) / np.sqrt(906044523.2)) if initial_val > 0 else 0
    })

# 4. Create the final DataFrame
df_summary = pd.DataFrame(results)

# Display the result
print(df_summary.to_string(index=False))
```

```python
Expected_events_2nd = 160000

df_summary["Error"] = ((df_summary["Events after 2nd Cut"] - Expected_events_2nd)/Expected_events_2nd)*100
```
